```python
# Brute force largest element search
arr = [1, 45, 54, 54, 4, 3, 2, 5, 6, 3,26, 74]
for i in range (0, len(arr)):
    for j in range (i + 1, len(arr) - 1):
        if arr[j] < arr[i]:
            arr[i],arr[j] = arr[j],arr[i]

print(arr)
print(f"Largest element is {arr[len(arr) - 1]}")
```

```python
# Optimal solution largest element search
arr = [1, 45, 54, 54, 4, 3, 2, 5, 6, 3,26, 74]
max = float('-inf')
for num in arr:
    if(num > max):
        max = num
print(f"Largest element is {max}")
```

```python
# Brute force second largest element search
arr = [1, 45, 54, 54, 4, 3, 2, 5, 6, 3,26, 74]
for i in range (0, len(arr)):
    for j in range (i + 1, len(arr) - 1):
        if arr[j] < arr[i]:
            arr[i],arr[j] = arr[j],arr[i]

print(arr)
max = arr[len(arr) - 1]
for i in range(len(arr) - 2, 0, -1):
    if (arr[i] != max):
        print(f"Second largest element is {arr[i]}")
        break
    else:
        print("No second largest element")
        break
```

```python
# Better solution second largest element search
arr = [1, 45, 54, 54, 4, 3, 2, 5, 6, 3,26, 74]
max = float('-inf')
for num in arr:
    if(num > max):
        max = num
print(f"Largest element is {max}")

smax = float('-inf')
for num in arr:
    if (num > smax and num != max):
        smax = num

if (smax == -1):
    print("No second largest element exists")
else:
    print(f"Second largest element is {smax}")
```

```python
# Optimal solution second largest and second smallest element search
arr = [1, 45, 54, 54, 4, 3, 2, 5, 6, 3,26, 74]
max = float('-inf')
smax = float('-inf')
for num in arr:
    if(num > max):
        smax = max
        max = num
    elif(num < max and num > smax):
        smax = num

min = arr[0]
smin = float('inf')

for num in arr:
    if(num < min):
        smin = min
        min = num
    elif(num != min and num < smin):
        smin = num

print(f"Largest element is {max}")
print(f"Smallest element is {min}")
print(f"Second largest element is {smax}")
print(f"Second smallest element is {smin}")
```

```python
# Brute force for unique elements
arr = [1, 1, 2, 2, 3, 3]
unique = []
k = 0

for num in arr:
    if num not in unique:
        unique.append(num)
        k += 1

for i in range(0, len(unique)):
    arr[i] = unique[i]

print(k)
print(arr)
```
