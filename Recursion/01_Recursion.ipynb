### RECURSION

```python
def test(i):
    return test(i + 1)

# This will cause a RecursionError: maximum recursion depth exceeded
# StackOverflow
```

```python
def test(i):
    if i >= 5:
        return

    print(i)
    test(i + 1)

test(0)
# This will print numbers from 0 to 4 using recursion and then stop
```

```python
i = 1
def display(i,n):
    if(i > n):
        return
    print("My name is Austin")

    display(i+1, n)

def main():
    print("Recursion program to print name multiple times")
    n = int(input("Enter the number of times to print the name: "))
    display(i,n)

main()
# This will print "My name is Austin" n times using recursion
```

```python
i = 1
def display1(i, n):
    if(i>n):
        return
    print(i)
    display1(i+1, n)

def main():
    print("Program to enter number from 1 to N linearly")
    n = int(input("Enter value of n: "))
    display1(i, n)

main()
# This will print numbers from 1 to n using recursion
```

```python
i = 1
def display2(i, n):
    if(n<i):
        return
    print(n)
    display2(i, n-1)

def main():
    print("Program to enter number from N to 1 linearly")
    n = int(input("Enter value of n: "))
    display2(i, n)

main()
# This will print numbers from n to 1 using recursion
```

```python
# Backtracking
def display3(i,n):
    if(i< 1 ):
        return
    display3(i-1, n)
    print(i)
def main():
    print("Program to enter number from 1 to N using backtracking")
    n = int(input("Enter value of n: "))
    display3(n, n)

main()
# This will print numbers from 1 to n using backtracking recursion
```

```python
i = 1
def display4(i, n):
    if(i>n):
        return
    display4(i+1, n)
    print(i)
def main():
    print("Program to enter from N to1 by backtracking")
    n = int(input("Enter value of N: "))
    display4(i,n)

main()
# This will print numbers from n to 1 using backtracking recursion
```

The time and space complexity for all the above will be O(n)
