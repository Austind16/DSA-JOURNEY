### RECURSION

```python
# Example of infinite recursion (will cause a stack overflow)
def test(i):
    return test(i + 1)

# This will cause a RecursionError: maximum recursion depth exceeded
# StackOverflow
```

```python
# Simple recursion example: print numbers from i to 4
def test(i):
    if i >= 5:  # Base case: stop recursion when i reaches 5
        return

    print(i)  # Print current value of i
    test(i + 1)  # Recursive call with incremented i

test(0)
# This will print numbers from 0 to 4 using recursion and then stop
```

```python
# Recursion to print a name multiple times
i = 1
def display(i,n):
    if(i > n):  # Base case: stop when i exceeds n
        return
    print("My name is Austin")  # Action: print the name

    display(i+1, n)  # Recursive call with incremented i

def main():
    print("Recursion program to print name multiple times")
    n = int(input("Enter the number of times to print the name: "))  # User input for n
    display(i,n)  # Start recursion

main()
# This will print "My name is Austin" n times using recursion
```

```python
# Recursion to print numbers from 1 to n (linear order)
i = 1
def display1(i, n):
    if(i>n):  # Base case: stop when i exceeds n
        return
    print(i)  # Print current value of i
    display1(i+1, n)  # Recursive call with incremented i

def main():
    print("Program to enter number from 1 to N linearly")
    n = int(input("Enter value of n: "))  # User input for n
    display1(i, n)  # Start recursion

main()
# This will print numbers from 1 to n using recursion
```

```python
# Recursion to print numbers from n to 1 (reverse order)
i = 1
def display2(i, n):
    if(n<i):  # Base case: stop when n is less than i
        return
    print(n)  # Print current value of n
    display2(i, n-1)  # Recursive call with decremented n

def main():
    print("Program to enter number from N to 1 linearly")
    n = int(input("Enter value of n: "))  # User input for n
    display2(i, n)  # Start recursion

main()
# This will print numbers from n to 1 using recursion
```

```python
# Backtracking recursion to print numbers from 1 to n
def display3(i,n):
    if(i< 1 ):  # Base case: stop when i is less than 1
        return
    display3(i-1, n)  # Recursive call with decremented i (goes down to 1)
    print(i)  # Print after recursion (backtracking)
def main():
    print("Program to enter number from 1 to N using backtracking")
    n = int(input("Enter value of n: "))  # User input for n
    display3(n, n)  # Start recursion from n

main()
# This will print numbers from 1 to n using backtracking recursion
```

```python
# Backtracking recursion to print numbers from n to 1
i = 1
def display4(i, n):
    if(i>n):  # Base case: stop when i exceeds n
        return
    display4(i+1, n)  # Recursive call with incremented i (goes up to n)
    print(i)  # Print after recursion (backtracking)
def main():
    print("Program to enter from N to1 by backtracking")
    n = int(input("Enter value of N: "))  # User input for n
    display4(i,n)  # Start recursion from i=1

main()
# This will print numbers from n to 1 using backtracking recursion
```

The time and space complexity for all the above will be O(n)
