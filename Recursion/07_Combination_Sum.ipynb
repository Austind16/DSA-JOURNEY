### Combination Sum I

```python
def combination(ind, target, arr, path, ans):
    # Recursive helper to find combinations that sum to `target` using elements from `arr`
    # `ind` is the current index in `arr` we are considering
    # `path` stores the current combination being built (mutable, per recursion branch)
    # `ans` collects all valid combinations found (list of lists)

    # Base case: if we've considered all indices
    if ind == len(arr):
        # If the remaining target is 0, record the current path as a valid combination
        if target == 0:
            # Append a copy to avoid later mutation when backtracking
            ans.append(path.copy())
        # Either way, no more indices to process in this branch
        return

    # Choice 1: pick the current element if it does not exceed the remaining target
    if arr[ind] <= target:
        # Include arr[ind] in the current path
        path.append(arr[ind])
        # Recurse without advancing index so that the same element may be reused
        combination(ind, target - arr[ind], arr, path, ans)
        # Backtrack: remove the last chosen element before exploring other choices
        path.pop()
    # Choice 2: skip the current element and move to the next index
    combination(ind + 1, target, arr, path, ans)

def main():
    # Example input array and target sum
    arr = [2, 3, 6, 7]
    target = 7
    # `ans` will collect all valid combinations (each combination is a list)
    ans = []
    # Start recursion with index 0 and an empty path
    combination(0, target, arr, [], ans)
    # Print all found combinations
    print(ans)

# Run the example
if __name__ == "__main__":
    main()
```
