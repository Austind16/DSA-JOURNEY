### Quick Sort

```python
# Program to perform Quick Sort
def qs(arr, low, high):
    # QuickSort recursive function: sorts arr[low..high]
    if low < high:
        # partition returns an index 'p' such that items <= pivot are on left
        pIndex = f(arr, low, high)
        qs(arr, low, pIndex)
        qs(arr, pIndex + 1, high)

def f(arr, low, high):
    pivot = arr[low]
    i = low - 1
    j = high + 1
    while True:
        # Move i right until arr[i] >= pivot
        i += 1
        while arr[i] < pivot:
            i += 1
        # Move j left until arr[j] <= pivot
        j -= 1
        while arr[j] > pivot:
            j -= 1
        # If pointers have crossed, partitioning is done
        if i >= j:
            return j
        # Otherwise swap elements at i and j
        arr[i], arr[j] = arr[j], arr[i]

def main():
    print("Program to perform Quick Sort")
    arr = [2, 1, 1, 5, 4, 4, 9, 6]
    l = len(arr)
    # pass last index (len-1) as high
    qs(arr, 0, l-1)
    print("Sorted array:", arr)

if __name__ == '__main__':
    main()
```
