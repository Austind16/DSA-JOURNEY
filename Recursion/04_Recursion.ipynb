### Subsequences problems

```python
# Recursive function to print all subsequences of an array whose sum equals 2
# ind: current index in the array
# arr: input array
# newarr: current subsequence being built
# sum: sum of elements in newarr

def subs(ind, arr, newarr, sum):

    n = len(arr)
    # Base case: if index reaches the end of the array

    if ind >= n:
        # If the sum of the current subsequence is 2, print it
        if sum == 2:
            print(newarr)
        return
    
    # Choice 1: Include the current element in the subsequence
    newarr.append(arr[ind])
    subs(ind + 1, arr, newarr, sum + arr[ind])

    # Backtrack: remove the last element to explore the next possibility
    newarr.pop()

    # Choice 2: Exclude the current element and move to the next
    subs(ind + 1, arr, newarr, sum)

# Main function to demonstrate the subsequence generation
def main():

    print("Program to print all subsequences of an array whose sum is 2:")

    arr = [1, 1, 2, 4, 2]
    
    # Start recursion with index 0, empty current subsequence, and sum 0
    subs(0, arr, newarr=[], sum=0)

main()
```

```python
# Function to print only one subsequence whose sum is 2
# Returns True as soon as one valid subsequence is found

def subs1(ind, arr, new_arr, curr_sum):
    n = len(arr)
    # Base case: if index reaches the end
    if ind >= n:
        if curr_sum == 2:
            # Condition satisfied, print and return Tru#####e
            print(new_arr)
            return True
        # Condition not satisfied
        return False
    # Include the current element
    new_arr.append(arr[ind])
    if subs1(ind + 1, arr, new_arr, curr_sum + arr[ind]):
        return True
    # Backtrack: remove the last element
    new_arr.pop()
    # Exclude the current element
    if subs1(ind + 1, arr, new_arr, curr_sum):
        return True
    # No valid subsequence found in this path
    return False

# Main function to demonstrate printing only one valid subsequence
def main():
    print("Program to print only one subsequence having sum 2:")
    arr = [1, 1, 2, 4, 2]
    subs1(0, arr, new_arr=[], curr_sum=0)

main()
```

```python
# Function to count all subsequences whose sum is 2
# Returns the count of valid subsequences

def subs1(ind, arr, new_arr, curr_sum):
    n = len(arr)
    # Base case: if index reaches the end
    if ind >= n:
        if curr_sum == 2:
            # Condition satisfied, print the valid subsequence
            print(new_arr)
            return 1
        # Condition not satisfied
        return 0
    # Include the current element
    new_arr.append(arr[ind])
    count_incl = subs1(ind + 1, arr, new_arr, curr_sum + arr[ind])
    # Backtrack: remove the last element
    new_arr.pop()
    # Exclude the current element
    count_excl = subs1(ind + 1, arr, new_arr, curr_sum)
    # Return total count from both choices
    return count_incl + count_excl

# Main function to demonstrate counting all valid subsequences
def main():
    print("Program to print and count all subsequences having sum 2:")
    arr = [1, 1, 2, 4, 2]
    total_count = subs1(0, arr, new_arr=[], curr_sum=0)
    print(f"Total valid subsequences: {total_count}")

main()
```
