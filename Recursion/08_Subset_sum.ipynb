### Subset Sum

```python
def subsum(ind, arr, current, result):

    # base case: we've considered every element
    if ind == len(arr):
        result.append(current)  # record the sum for this subset
        return

    # 1) Exclude the current element and move to next index
    subsum(ind + 1, arr, current, result)

    # 2) Include the current element and move to next index
    subsum(ind + 1, arr, current + arr[ind], result)


def main():
    # example input
    arr = [3, 1, 2]

    # start from index 0 with sum 0; `result` will collect all subset sums
    current = 0
    result = []
    subsum(0, arr, current, result)

    # `result` now contains sums for every subset (order: DFS traversal)
    print("All subset sums:", result)


if __name__ == "__main__":
    main()
```

```python
def subs(ind, arr, path, result):
    result.append(path.copy())

    for i in range(ind, len(arr)):
        if( i != ind and arr[i] == arr[i-1]):
            continue
        path.append(arr[i])
        subs(i + 1, arr, path, result)
        path.pop()

def main():
    arr = [1, 2, 2, 3, 3]
    result = []
    subs(0, arr, [], result)
    print(result)
 

if __name__ == "__main__":
    main()
```
