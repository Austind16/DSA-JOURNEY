### Multiple recursive call


```python
# Recursive function to calculate the nth Fibonacci number
def fibo(n):
    # Base case: if n is 0 or 1, return n (Fibonacci(0)=0, Fibonacci(1)=1)
    if (n <= 1):
        return n
    last = fibo(n-1)
    slast = fibo(n-2)
    return last + slast

def main():
    print("Program to print the nth Fibonacci number using recursion:")
    n = int(input("Enter value of n: "))  # User input for n
    print(f"The {n}th Fibonacci number is: {fibo(n)}")  # Print the nth Fibonacci number

main()
```

```python
# Recursive function to print the Fibonacci sequence up to n terms
def fibo(n):
    # Base case: Return n if n is 0 or 1 (handles the first two Fibonacci numbers)
    if (n <= 1):
        return n
    # Recursive case: Calculate the sum of the previous two Fibonacci numbers
    last = fibo(n-1)
    slast = fibo(n-2)
    return last + slast

def main():
    print("Program to print the Fibonacci sequence up to n terms:")
    n = int(input("Enter value of n: "))  # Get the number of terms from the user
    for i in range(n):
        print(fibo(i))  # Print the ith Fibonacci number in the sequence

main()
```

### Recursion on subsequences

```python
# Function to print all subsequences of an array using recursion
def subs(ind, arr, newarr):
    n = len(arr)
    # Base case: if index reaches the end, print the current subsequence
    if(ind >= n):
        print(newarr)
        return 
    # Include the current element in the subsequence
    newarr.append(arr[ind])
    subs(ind + 1, arr, newarr)
    # Backtrack: remove the last element and explore the path without including it
    newarr.pop()
    subs(ind + 1, arr, newarr)

def main():
    print("Program to print all subsequences of an array: ")
    arr = [3, 1, 2]
    # Start recursion with index 0 and an empty list for the current subsequence
    subs(0, arr, newarr = [])

main()
```
