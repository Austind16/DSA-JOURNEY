### Parameterised and functional recursion

```python
# Parameterized recursion to calculate the sum of first N numbers
sum = 0 
def f(i, sum):
    if(i<1):  # Base case: when i is less than 1, print the sum and stop
        print(f"Sum is: {sum}")
        return
    f(i - 1, sum + i)  # Recursive call with i-1 and updated sum
def main():
    print("Program to get sum of N numbers using parameterised recursion: ")
    n = int(input("Enter number till which you want sum: "))  # User input for n
    f(n, sum)  # Start recursion

main()
# Program to get sum of N numbers using Parameterized recursion
```

```python
# Functional recursion to calculate the sum of first N numbers
def f1(n):
    if(n == 0):  # Base case: sum of 0 is 0
        return 0
    return n + f1(n-1)  # Recursive call and sum up
def main():
    print("Program to get sum of first N numbers using functional recursion: ")
    n = int(input("Enter number till which you want sum: "))  # User input for n
    print(f1(n))  # Print the result

main()
# Program to get sum of N numbers using Functional recursion
```

```python
# Functional recursion to calculate factorial of a number
def f2(n):
    if(n == 1 or n == 0):  # Base case: factorial of 0 or 1 is 1
        return 1
    return n * f2(n-1)  # Recursive call and multiply
def main():
    print("Program to get factorial of a number using functional recursion: ")
    n = int(input("Enter number which you want factorial of: "))  # User input for n
    print(f2(n))  # Print the result

main()
```

```python
# Parameterized recursion to calculate factorial of a number
product = 1
def f3(n, product):
    if(n == 1 or n == 0):  # Base case: factorial of 0 or 1 is 1
        print(f"Product is: {product}")
        return
    f3(n-1, product * n)  # Recursive call with n-1 and updated product

def main():
    print("Program to find factorial of a number using parameterised recursion: ")
    n = int(input("Enter number which you want to find factorial of: "))  # User input for n
    f3(n, product)  # Start recursion

main()
```

```python
# Recursion to reverse an array using two pointers (left and right)
import numpy as np
def f4(arr, l, r):
    if(l >= r):  # Base case: when left index crosses right, stop
        return
    arr[l], arr[r] = arr[r], arr[l]  # Swap elements at l and r
    f4(arr, l + 1, r - 1)  # Recursive call with updated pointers
def main():
    print("Program to reverse an array: ")
    arr = np.array([1,2,3,6,9,2])
    l = 0
    r = len(arr) - 1
    f4(arr, l, r)  # Start recursion
    print("Reversed array: ", arr)
main()
```

```python
# Recursion to reverse an array using a single pointer (left index)
def f5(arr, l):
    n = len(arr)
    if(l >= n//2):  # Base case: when left index reaches the middle, stop
        return
    arr[l], arr[n-1-l] = arr[n-1-l], arr[l]  # Swap elements at l and its mirror index
    f5(arr, l + 1)  # Recursive call with incremented left pointer
def main():
    print("Program to reverse an array: ")
    arr = np.array([1,2,3,6,9,2])
    f5(arr, 0)  # Start recursion
    print("Reversed array: ", arr)

main()
```

```python
# Recursion to check if a string is a palindrome
def f6(s, l):
    n = len(s)
    if(l >= n/2):  # Base case: checked all pairs, string is a palindrome
        return True
    if(s[l] != s[n-l-1]):  # If mismatch found, not a palindrome
        return False
    return f6(s, l+1)  # Recursive call to check next pair
def main():
    print("Program to check for palindrome: ")
    s = input("Enter string to check: ")  # User input for string
    print(f6(s, 0))  # Print result (True/False)

main()
# Program to check for palindrome using parameterised recursion
```
