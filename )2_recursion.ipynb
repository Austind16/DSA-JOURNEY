### Parameterised and functional recursion

```python
sum = 0 
def f(i, sum):
    if(i<1):
        print(f"Sum is: {sum}")
        return
    f(i - 1, sum + i)
def main():
    print("Program to get sum of N numbers using parameterised recursion: ")
    n = int(input("Enter number till which you want sum: "))
    f(n, sum)

main()
# Program to get sum of N numbers using Parameterized recursion
```

```python
def f1(n):
    if(n == 0):
        return 0
    return n + f1(n-1)
def main():
    print("Program to get sum of first N numbers using functional recursion: ")
    n = int(input("Enter number till which you want sum: "))
    print(f1(n))

main()
# Program to get sum of N numbers using Functional recursion
```

```python
def f2(n):
    if(n == 1 or n == 0):
        return 1
    return n * f2(n-1)
def main():
    print("Program to get factorial of a number using functional recursion: ")
    n = int(input("Enter number which you want factorial of: "))
    print(f2(n))

main()
```

```python
product = 1
def f3(n, product):
    if(n == 1 or n == 0):
        print(f"Product is: {product}")
        return
    f3(n-1, product * n)

def main():
    print("Program to find factorial of a number using parameterised recursion: ")
    n = int(input("Enter number which you want to find factorial of: "))
    f3(n, product)

main()
```

```python
import numpy as np
def f4(arr, l, r):
    if(l >= r):
        return
    arr[l], arr[r] = arr[r], arr[l]
    f4(arr, l + 1, r - 1)
def main():
    print("Program to reverse an array: ")
    arr = np.array([1,2,3,6,9,2])
    l = 0
    r = len(arr) - 1
    f4(arr, l, r)
    print("Reversed array: ", arr)
main()
```

```python
def f5(arr, l):
    n = len(arr)
    if(l >= n//2):
        return
    arr[l], arr[n-1-l] = arr[n-1-l], arr[l]
    f5(arr, l + 1)
def main():
    print("Program to reverse an array: ")
    arr = np.array([1,2,3,6,9,2])
    f5(arr, 0)
    print("Reversed array: ", arr)

main()
```

```python
def f6(s, l):
    n = len(s)
    if(l >= n/2):
        return True
    if(s[l] != s[n-l-1]):
        return False
    return f6(s, l+1)
def main():
    print("Program to check for palindrome: ")
    s = input("Enter string to check: ")
    print(f6(s, 0))

main()
# Program to check for palindrome using parameterised recursion
```
